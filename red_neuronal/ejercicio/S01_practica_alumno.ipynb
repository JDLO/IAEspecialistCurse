## De kilómetros a millas
Enseña a una red a pasar de **km** a **millas**.

👉 Copia del cuaderno de ejemplo y cambia lo necesario.

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras

X = np.array([1, 5, 10, 21, 42, 50, 100], dtype=float)           # km
y = np.array([0.6, 3.1, 6.2, 13, 26.1, 31, 62.1], dtype=float)   # millas
```

**1.** Dibuja los ejemplos.

```python
# ✏️ TU CÓDIGO AQUÍ
plt.scatter(X, y)
plt.plot(X,y, linestyle="--", color='r')
plt.ylabel("Millas")
plt.xlabel("Kilometros")
plt.show()
```

**2.** Crea la red (igual que en el ejemplo).

```python
# ✏️ TU CÓDIGO AQUÍ
modelo = keras.Sequential([
    keras.Input(shape=(1,)),
    keras.layers.Dense(1, 
                    #    kernel_initializer=keras.initializers.GlorotUniform(seed=42) # Make for added a seed for the first moment to try the weights
                       )
])

modelo.compile(loss="mean_squared_error",
               optimizer=keras.optimizers.Adam(0.1))
```

**3.** Entrena la red **300 veces** (`epochs=300`).

```python
# ✏️ TU CÓDIGO AQUÍ
conversor = modelo.fit(X, y, epochs=300, verbose=0)
print("Entrenado!")
```

**4.** Dibuja el error. ¿Baja?

```python
# ✏️ TU CÓDIGO AQUÍ
plt.plot(conversor.history["loss"])
plt.xlabel("Repeticiones")
plt.ylabel("Error")
plt.show()
```

**5.** ¿Cuántas millas son **42 km**? La respuesta correcta es unas **26**.

```python
# ✏️ TU CÓDIGO AQUÍ
kilometro = np.array([42], dtype=float)
print(f"Cuantas millas son {kilometro[0]} Km ? : {modelo.predict(kilometro)[0][0]:,.2f}")
```

**6.** Mira el peso y el sesgo. La regla real es **millas = km × 0.62**. ¿Se parece el peso a 0.62?

```python
# ✏️ TU CÓDIGO AQUÍ
peso, sesgo = modelo.get_weights()
print(f"El peso es de {peso[0][0]:,.3f}")
print(f"El sesgo es de {sesgo[0]:,.3f}")
```
