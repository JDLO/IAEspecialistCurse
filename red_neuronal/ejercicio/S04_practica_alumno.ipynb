## El precio de un coche de segunda mano
Predice el precio de un coche sabiendo sus **años** y sus **kilómetros**.

👉 Completa donde pone `# ✏️ TU CÓDIGO AQUÍ`. Copia del cuaderno de ejemplo y cambia lo necesario.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow import keras
from sklearn.model_selection import train_test_split
```

```python
# Esta celda crea el CSV (no hace falta cambiarla)
np.random.seed(1)
anios = np.random.randint(0, 15, 200)
km = np.random.randint(0, 200, 200)            # en miles de km
precio = 25 - 1.2 * anios - 0.04 * km + np.random.normal(0, 1, 200)   # en miles de €
filename = './csv/coches.csv'
pd.DataFrame({"anios": anios, "km": km, "precio": precio.round(1)}).to_csv(filename, index=False)
```

**1.** Lee el CSV `coches.csv` y muestra las primeras filas.

```python
# ✏️ TU CÓDIGO AQUÍ
data = pd.read_csv(filename)
data.head()
```

**2.** Separa X (`anios` y `km`) e y (`precio`).

```python
# ✏️ TU CÓDIGO AQUÍ
X = data[["anios", "km"]]
y = data["precio"]

display(X.info())
display(y.info())
```

**3.** Separa train y test (20 % para el examen).

```python
# ✏️ TU CÓDIGO AQUÍ
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
print("Datos de entrenamiento: ", len(X_train))
print("Datos de examen: ", len(X_test))
```

**4.** Crea la red: 2 entradas, 2 capas ocultas de 16 con `relu` y 1 salida. Compílala con `Adam(0.01)`.

```python
# ✏️ TU CÓDIGO AQUÍ
red = keras.Sequential([
    keras.Input(shape=(2,)),
    keras.layers.Dense(32, activation="relu"),
    keras.layers.Dense(32, activation="relu"),
    keras.layers.Dense(32, activation="relu"),
    keras.layers.Dense(1),
])
red.compile(loss="mean_squared_error",
            optimizer=keras.optimizers.Adam(0.013)
)
```

**5.** Entrena 200 épocas y dibuja el error.

```python
# ✏️ TU CÓDIGO AQUÍ
mTrain = red.fit(X_train, y_train, epochs=600, verbose=0)
print("Entrenamiento culminado!!")
```

```python
plt.plot(mTrain.history["loss"])
plt.xlabel("Repeticiones")
plt.ylabel("Error")
plt.show()
```

**6.** Haz el examen: calcula el error medio.

```python
# ✏️ TU CÓDIGO AQUÍ
prediccion = red.predict(X_test, verbose=0).flatten()

resultado = pd.DataFrame({"precio real": y_test.values, "precio red": prediccion.round(1)})
resultado.head(10)

error_medio = np.mean(np.abs(y_test.values - prediccion))
print("La red se equivoca de media en", round(error_medio, 1), "mil €")
```

```python
plt.scatter(y_test, prediccion)
limites = [
        min(y_test.min(), prediccion.min()),
        max(y_test.max(), prediccion.max())
    ]
plt.plot(limites, limites, color="red")
plt.xlabel("Precio real")
plt.ylabel("Precio de la red")
plt.show()
```

**7.** ¿Cuánto vale un coche de **5 años** y **80 mil km**?

```python
# ✏️ TU CÓDIGO AQUÍ
coche = pd.DataFrame({"anios": [5], "km": [80.0]})
print(red.predict(coche, verbose=0), "mil €")
```

MAE y R2

```python
from sklearn.metrics import mean_absolute_error, r2_score
mae = mean_absolute_error(y_test, prediccion)
r2 = r2_score(y_test, prediccion)

print(f"Error medio (MAE) {mae:,.3f}")
print(f"r2  {r2:,.3f}")
```
