# 🧠 Sesión 1 · Mi primera red neuronal

Una **red neuronal** es un programa que **aprende con ejemplos**.

Hoy le enseñaremos a pasar de **grados Celsius** a **grados Fahrenheit**.
No le diremos cómo se hace: solo le daremos **ejemplos**.

## 1. Preparar Colab

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
```

## 2. Los ejemplos
- **X** = la pregunta (Celsius)
- **y** = la respuesta correcta (Fahrenheit)

```python
X = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)
y = np.array([-40, 14, 32, 46.4, 59, 71.6, 100.4], dtype=float)

plt.scatter(X, y)
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")
plt.show()
```

## 3. Crear la red
Nuestra red tiene **1 neurona**.

💡 Estas líneas las copiaremos **siempre casi iguales**. No hace falta entender cada palabra.

```python
modelo = keras.Sequential([
    keras.Input(shape=(1,)),
    keras.layers.Dense(1)
])

modelo.compile(loss="mean_squared_error",
               optimizer=keras.optimizers.Adam(0.1))
```

## 4. Entrenar la red
La red mira los ejemplos **500 veces** y va aprendiendo.

```python
historial = modelo.fit(X, y, epochs=500, verbose=0)
print("¡Red entrenada! ✅")
```

## 5. ¿Ha aprendido?
Esta gráfica muestra **cuánto se equivoca** la red.
Si la línea **baja**, la red está aprendiendo.

```python
plt.plot(historial.history["loss"])
plt.xlabel("Repeticiones")
plt.ylabel("Error")
plt.show()
```

## 6. Usar la red
¿Cuántos Fahrenheit son **100 °C**? La respuesta correcta es **212**.

```python
print(modelo.predict(np.array([100.0]), verbose=0))
```

🎉 **¡La red ha aprendido sola!** Nadie le ha dicho cómo se hace.

## 7. ¿Qué ha aprendido por dentro?
Una neurona solo guarda **2 números**:
- el **peso** → por cuánto **multiplica**
- el **sesgo** → cuánto **suma**

La regla real es: **Fahrenheit = Celsius × 1.8 + 32**

Vamos a ver qué números ha encontrado la red:

```python
peso, sesgo = modelo.get_weights()
print("Peso: ", peso)    # debería salir cerca de 1.8
print("Sesgo:", sesgo)   # debería salir cerca de 32
```

✅ ¡Ha encontrado el **1.8** y el **32** ella sola!

💡 **Aprender** = buscar los pesos y sesgos que hacen que la red acierte.

## 📝 Resumen
Para hacer una red siempre seguimos estos pasos:
1. **Ejemplos** → `X` e `y`
2. **Crear** la red
3. **Entrenar** → `fit`
4. **Usar** → `predict`
