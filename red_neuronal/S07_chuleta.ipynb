### ¿Qué plantilla uso?
- ¿Quiero predecir **un número** (precio, nota, temperatura...)? → **Plantilla A**
- ¿Quiero predecir **sí o no**? → **Plantilla B**
- ¿Quiero elegir entre **varias opciones**? → **Plantilla C**

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow import keras
from sklearn.model_selection import train_test_split
````

## Pasos comunes (siempre iguales)
```python
datos = pd.read_csv("mi_archivo.csv")
X = datos[["columna1", "columna2"]]     # lo que sé
y = datos["columna_respuesta"]          # lo que quiero predecir
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

---
## Plantilla A · Predecir un número
```python
modelo = keras.Sequential([
    keras.Input(shape=(NUM_COLUMNAS,)),
    keras.layers.Dense(16, activation="relu"),
    keras.layers.Dense(16, activation="relu"),
    keras.layers.Dense(1)
])
modelo.compile(loss="mean_squared_error", optimizer=keras.optimizers.Adam(0.01))
modelo.fit(X_train, y_train, epochs=200, verbose=0)

prediccion = modelo.predict(X_test, verbose=0).flatten()
print("Error medio:", np.mean(np.abs(y_test.values - prediccion)))
```

---
## Plantilla B · Predecir sí o no
```python
modelo = keras.Sequential([
    keras.Input(shape=(NUM_COLUMNAS,)),
    keras.layers.Dense(16, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid")
])
modelo.compile(loss="binary_crossentropy", optimizer=keras.optimizers.Adam(0.01), metrics=["accuracy"])
modelo.fit(X_train, y_train, epochs=100, verbose=0)

perdida, aciertos = modelo.evaluate(X_test, y_test, verbose=0)
print("Aciertos:", aciertos)
```

---
## Plantilla C · Elegir entre varias opciones
```python
modelo = keras.Sequential([
```

## ⚠️ Errores típicos
| Error | Solución |
|---|---|
| La red no aprende (el error no baja) | Pon más `epochs` o prueba `Adam(0.1)` |
| `shape` no coincide | En `keras.Input(shape=(...,))` pon el **número de columnas** de X |
| En sí/no salen números raros | ¿Has puesto `activation="sigmoid"` en la última capa? |
| Las columnas tienen números muy distintos (0.1 y 1000) | Usa `StandardScaler` como en la sesión 6 |
