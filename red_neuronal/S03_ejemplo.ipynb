# 🧠 Sesión 3 · Rectas y curvas

Una neurona sola solo sabe dibujar **líneas rectas**.

Pero en la vida real muchas cosas son **curvas**. Hoy veremos cómo hacer que la red aprenda curvas.

````python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
````

## Los datos: la altura de una pelota
Lanzamos una pelota hacia arriba: **sube y luego baja**. Es una curva.

````python
segundos = np.linspace(0, 6, 60)
altura = -(segundos - 3) ** 2 + 9

plt.scatter(segundos, altura)
plt.xlabel("Segundos")
plt.ylabel("Altura (m)")
plt.show()
````

## Red 1 · Una sola neurona
Como en las sesiones anteriores.

````python
red1 = keras.Sequential([
    keras.Input(shape=(1,)),
    keras.layers.Dense(1)
])
red1.compile(loss="mean_squared_error", optimizer=keras.optimizers.Adam(0.01))
red1.fit(segundos, altura, epochs=500, verbose=0)

plt.scatter(segundos, altura, color="gray")
plt.plot(segundos, red1.predict(segundos, verbose=0), color="red")
plt.title("Red de 1 neurona")
plt.show()
````

❌ Solo sabe hacer una **recta**. No sirve para la pelota.

## Red 2 · Más neuronas y la palabra mágica: `relu`
Hacemos 2 cambios:
1. Ponemos **capas ocultas** con más neuronas: `Dense(16)`.
2. Les añadimos `activation="relu"`.

💡 `relu` es lo que permite a la red **doblar** la línea. Sin `relu`, aunque pongas muchas neuronas, sigue haciendo una recta.

````python
seed = keras.initializers.GlorotUniform(seed=42)
red2 = keras.Sequential([
    keras.Input(shape=(1,)),
    keras.layers.Dense(64, activation="relu"),   # capa oculta
    keras.layers.Dense(64, activation="tanh"),   # capa oculta
    keras.layers.Dense(64, activation="relu"),   # capa oculta
    keras.layers.Dense(1)                        # salida
])
red2.compile(loss="mean_squared_error", optimizer=keras.optimizers.Adam(0.03))
conversor = red2.fit(segundos, altura, epochs=200, verbose=0)

plt.scatter(segundos, altura, color="gray")
plt.plot(segundos, red2.predict(segundos, verbose=0), color="green")
plt.title("Red con capas ocultas y relu")
plt.show()
````

````python
# ✏️ TU CÓDIGO AQUÍ
plt.plot(conversor.history["loss"])
plt.xlabel("Repeticiones")
plt.ylabel("Error")
plt.show()
````

✅ ¡Ahora sí sigue la curva!

## Así es la red que hemos hecho
```
 Entrada  →  16 neuronas  →  16 neuronas  →  Salida
(segundos)     (relu)          (relu)        (altura)
```
Las capas del medio se llaman **capas ocultas**.

## Usamos la red
¿A qué altura está la pelota a los **2 segundos**? (La respuesta correcta es 8 m.)

````python
print(red2.predict(np.array([2.0]), verbose=0))
````

## 📝 Resumen
- **1 neurona** → solo rectas.
- **Capas ocultas + `relu`** → curvas.
- A partir de ahora, las capas ocultas **siempre** llevarán `activation="relu"`.
