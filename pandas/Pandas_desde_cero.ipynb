## 1. ¿Qué es pandas y para qué sirve?

**pandas** es la librería de Python para trabajar con **datos en forma de tabla**, igual que en una hoja de Excel: filas, columnas, nombres de columnas...

Con pandas podemos:
- **Leer** datos de archivos (CSV, Excel...).
- **Explorarlos** (ver las primeras filas, estadísticas, tipos de datos...).
- **Limpiarlos** (datos que faltan, filas repetidas...).
- **Filtrarlos, ordenarlos y resumirlos**.

Es el paso previo imprescindible antes de aplicar Machine Learning: **los datos casi nunca llegan limpios y listos**.

Pandas tiene dos tipos de objetos principales:

| Objeto | Qué es | Parecido en Excel |
|--------|--------|-------------------|
| `Series` | Una **columna** de datos | Una columna |
| `DataFrame` | Una **tabla** completa (varias columnas) | Una hoja entera |

Para usarla, primero se **importa**. Por costumbre se le pone el apodo `pd` (y también importamos NumPy como `np`, que usaremos alguna vez):

```python
import pandas as pd
import numpy as np

print("pandas importado correctamente")
print("Versión:", pd.__version__)
```

## 2. Series: una columna de datos

Una **Serie** es como un array de NumPy, pero con una **etiqueta** (índice) para cada valor. Se crea con `pd.Series()` pasándole una lista:

```python
temperaturas = pd.Series([18, 21, 19, 24, 22])
print(temperaturas)
```

Fíjate en que hay **dos columnas**: a la izquierda el **índice** (0, 1, 2...) y a la derecha los **valores**. Abajo aparece el tipo de dato (`dtype`).

Podemos poner **nuestras propias etiquetas** de índice, y así acceder a cada valor por su nombre:

```python
temp = pd.Series([18, 21, 19, 24, 22],
                 index=["Lun", "Mar", "Mié", "Jue", "Vie"])
print(temp)
print()
print("Temperatura del miércoles:", temp["Mié"])
print("Temperatura media:", temp.mean())
```

Las Series admiten las mismas operaciones y filtros que los arrays de NumPy:

```python
print(temp + 1)                 # Suma 1 a todos los valores
print()
print(temp[temp > 20])          # Solo los días de más de 20 grados
```

## 3. DataFrame: una tabla de datos

Un **DataFrame** es una tabla: varias Series (columnas) juntas, que comparten el mismo índice.

La forma más sencilla de crearlo es con un **diccionario**: cada **clave** es el nombre de una columna y cada **valor** es la lista de datos de esa columna.

Vamos a crear la tabla de notas de un grupo de alumnos. La usaremos durante todo el cuaderno:

```python
alumnos = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Pablo", "Sara", "Diego"],
    "edad": [20, 22, 21, 23, 20, 22],
    "ciudad": ["Madrid", "Sevilla", "Madrid", "Valencia", "Sevilla", "Madrid"],
    "mates": [7.5, 5.0, 9.0, 4.5, 6.0, 8.0],
    "lengua": [6.5, 7.0, 8.5, 5.0, 7.5, 6.0],
})

alumnos
```

La columna de la izquierda (0, 1, 2...) es el **índice**: pandas lo crea automáticamente. Cada **fila** es un alumno y cada **columna** es un dato.

### 3.1 Explorar un DataFrame

Estas son las instrucciones que más se usan al empezar a trabajar con una tabla:

| Instrucción | Qué muestra |
|-------------|-------------|
| `df.head(n)` | Las primeras `n` filas (por defecto 5) |
| `df.tail(n)` | Las últimas `n` filas |
| `df.shape` | `(filas, columnas)` |
| `df.columns` | Los nombres de las columnas |
| `df.dtypes` | El tipo de dato de cada columna |
| `df.info()` | Resumen: columnas, tipos y valores no nulos |
| `df.describe()` | Estadísticas de las columnas numéricas |

```python
print("Primeras 3 filas:")
print(alumnos.head(3))

print("\nÚltimas 2 filas:")
print(alumnos.tail(2))

print("\nForma (filas, columnas):", alumnos.shape)
print("Columnas:", list(alumnos.columns))
```

```python
alumnos.info()
```

En `info()` se ve cuántos valores **no nulos** hay en cada columna (aquí 6 en todas: no falta ningún dato) y el tipo (`int64` = enteros, `float64` = decimales, `object`/`str` = texto).

```python
alumnos.describe()
```

`describe()` calcula para cada columna numérica: `count` (cuántos datos), `mean` (media), `std` (desviación típica), `min`, los cuartiles (`25%`, `50%` = mediana, `75%`) y `max`. **Es el primer vistazo rápido que se da a unos datos nuevos.**

## 4. Leer y guardar archivos CSV

En la vida real los datos vienen en **archivos**. El formato más habitual es el **CSV** (*Comma-Separated Values*): un archivo de texto donde cada fila es una línea y los valores se separan con comas.

## 5. Seleccionar columnas y filas

### 5.1 Seleccionar columnas

- Con **un par de corchetes** y un nombre, obtenemos **una columna** (una Serie).
- Con **dos pares de corchetes** y una lista de nombres, obtenemos **varias columnas** (un DataFrame).

```python
print(df["nombre"])
```

```python
df[["nombre", "mates"]]
```

### 5.2 Seleccionar filas por posición: `iloc`

`iloc` (*integer location*) selecciona por **posición numérica**, empezando en 0, igual que en NumPy: `df.iloc[fila, columna]`.

```python
print(df.iloc[0])             # La primera fila
print()
print(df.iloc[2, 1])          # Fila 2, columna 1
```

```python
df.iloc[:3]                   # Las 3 primeras filas
```

```python
df.iloc[:3, [0, 3]]           # 3 primeras filas, columnas 0 y 3
```

### 5.3 Seleccionar por etiqueta: `loc`

`loc` selecciona usando **etiquetas** (los nombres del índice y de las columnas). Para verlo mejor, vamos a usar el **nombre** del alumno como índice con `set_index`:

```python
por_nombre = df.set_index("nombre")
por_nombre
```

```python
print(por_nombre.loc["Ana"])                       # Toda la fila de Ana
print()
print(por_nombre.loc["Marta", "mates"])            # Nota de mates de Marta
```

```python
por_nombre.loc[["Ana", "Sara"], ["mates", "lengua"]]
```

> 💡 **Resumen:** `iloc` = por **posición** (0, 1, 2...). `loc` = por **etiqueta** (los nombres).
>
> 💡 `set_index` **no modifica** `df`: devuelve un DataFrame nuevo (por eso lo hemos guardado en `por_nombre`). Esto es habitual en pandas: casi todas las operaciones devuelven un resultado nuevo.

## 6. Filtrar datos con condiciones

Igual que en NumPy: una condición produce un conjunto de `True`/`False`, y si lo ponemos entre corchetes nos quedamos solo con las filas `True`.

Primero, veamos la condición por sí sola:

```python
print(df["mates"] >= 5)
```

Ahora la usamos como filtro: **alumnos que aprueban matemáticas**:

```python
df[df["mates"] >= 5]
```

Podemos **contar** cuántas filas cumplen la condición con `len()`:

```python
aprobados_mates = df[df["mates"] >= 5]
print("Aprueban mates:", len(aprobados_mates), "de", len(df))
```

### Varias condiciones a la vez

Se combinan con `&` (**y**), `|` (**o**) y `~` (**no**), y **cada condición va entre paréntesis**:

```python
# Alumnos de Madrid que han sacado al menos un 7 en mates
df[(df["ciudad"] == "Madrid") & (df["mates"] >= 7)]
```

```python
# Alumnos que suspenden mates O suspenden lengua
df[(df["mates"] < 5) | (df["lengua"] < 5)]
```

### Otros filtros útiles

- `isin([...])`: comprueba si el valor está dentro de una lista.
- `str.startswith(...)`: comprueba cómo empieza un texto.

```python
# Alumnos de Sevilla o Valencia
df[df["ciudad"].isin(["Sevilla", "Valencia"])]
```

```python
# Alumnos cuyo nombre empieza por M
df[df["nombre"].str.startswith("M")]
```

## 7. Crear y modificar columnas

### 7.1 Crear una columna nueva

Se hace como si fuera un diccionario: `df["nombre_nueva"] = ...`. Vamos a calcular la **nota media** de cada alumno:

```python
df["media"] = (df["mates"] + df["lengua"]) / 2
df
```

Gracias a que pandas opera columna a columna, **no hace falta ningún bucle**: la operación se aplica a todas las filas a la vez.

### 7.2 Columna según una condición (`np.where`)

Igual que con NumPy: `np.where(condición, valor_si_True, valor_si_False)`:

```python
df["resultado"] = np.where(df["media"] >= 5, "Aprobado", "Suspenso")
df
```

### 7.3 Modificar valores concretos

Con `loc` podemos cambiar valores de las filas que cumplan una condición: `df.loc[condición, "columna"] = nuevo_valor`.

Imaginemos que Pablo ha recuperado matemáticas y ahora tiene un 5.0. Cambiamos su nota y **recalculamos** la media y el resultado:

```python
df.loc[df["nombre"] == "Pablo", "mates"] = 5.0

df["media"] = (df["mates"] + df["lengua"]) / 2
df["resultado"] = np.where(df["media"] >= 5, "Aprobado", "Suspenso")
df
```

### 7.4 Renombrar y eliminar columnas

- `rename(columns={"viejo": "nuevo"})` cambia nombres.
- `drop(columns=["col"])` elimina columnas.

Ambas **devuelven un DataFrame nuevo** (el original no cambia, salvo que lo reasignemos):

```python
resumen = df.rename(columns={"mates": "nota_mates", "lengua": "nota_lengua"})
resumen = resumen.drop(columns=["edad", "ciudad"])
resumen
```

## 8. Ordenar y calcular estadísticas

### 8.1 Ordenar con `sort_values`

```python
# Ordenados por media, de mayor a menor
df.sort_values("media", ascending=False)
```

Para quedarnos con los **3 mejores**, encadenamos `head(3)`:

```python
df.sort_values("media", ascending=False).head(3)
```

### 8.2 Estadísticas de una columna

Cada columna (Serie) tiene métodos como `mean()`, `median()`, `max()`, `min()`, `sum()` y `count()`:

```python
print("Media de mates   :", df["mates"].mean())
print("Mediana de mates :", df["mates"].median())
print("Nota máxima      :", df["mates"].max())
print("Nota mínima      :", df["mates"].min())
print("Suma de edades   :", df["edad"].sum())
```

También se puede calcular sobre **varias columnas** a la vez:

```python
df[["mates", "lengua"]].mean()
```

Y para saber **quién** tiene el valor máximo, usamos `idxmax()` (devuelve el índice de la fila) junto con `loc`:

```python
fila_mejor = df["media"].idxmax()
print("Mejor media:", df.loc[fila_mejor, "nombre"], "con", df.loc[fila_mejor, "media"])
```

### 8.3 Contar categorías: `value_counts`

Para columnas de texto o categorías, `value_counts()` cuenta cuántas veces aparece cada valor:

```python
df["ciudad"].value_counts()
```

### 8.4 Estadísticas por grupos: `groupby`

Una de las herramientas más potentes de pandas: **dividir la tabla en grupos**, calcular algo en cada grupo y juntar los resultados.

Por ejemplo, **la nota media de mates y lengua en cada ciudad**:

```python
df.groupby("ciudad")[["mates", "lengua"]].mean()
```

Se lee así: *"agrupa `df` por `ciudad` y, de las columnas `mates` y `lengua`, calcula la media"*. Se puede cambiar `mean()` por `sum()`, `max()`, `min()`, `count()`...

Y con `round()` limpiamos los decimales:

```python
df.groupby("ciudad")[["mates", "lengua"]].mean().round(2)
```

## 9. Datos que faltan y filas repetidas

En datos reales es muy habitual que haya **huecos** (valores que faltan) y **filas duplicadas**. Vamos a crear una tabla "sucia" a propósito. Cuando falta un dato, pandas lo representa como **`NaN`** (*Not a Number*):

```python
sucio = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Luis", "Sara"],
    "edad": [20, None, 22, None, 21],
    "nota": [7.5, 6.0, None, 6.0, 5.5],
})

sucio
```

### 9.1 Detectar valores que faltan

`isnull()` marca con `True` los huecos, y con `.sum()` los contamos **por columna**:

```python
print(sucio.isnull().sum())
```

### 9.2 Opción A: eliminar las filas con huecos (`dropna`)

```python
sucio.dropna()
```

Es rápido, pero **perdemos datos**: aquí hemos pasado de 5 filas a 2.

### 9.3 Opción B: rellenar los huecos (`fillna`)

Podemos sustituir los huecos por un valor, por ejemplo **la media** de la columna:

```python
relleno = sucio.copy()
relleno["nota"] = relleno["nota"].fillna(relleno["nota"].mean())
relleno["edad"] = relleno["edad"].fillna(relleno["edad"].median())
relleno
```

> 📝 `sucio.copy()` crea una copia independiente para no modificar la tabla original. 
>
> 📝 Qué opción elegir (borrar o rellenar, y con qué valor) depende de los datos. Lo veremos con más detalle en el preprocesado de Machine Learning.

### 9.4 Filas duplicadas

`duplicated()` marca las filas que ya han aparecido antes, y `drop_duplicates()` las elimina:

```python
print("Filas duplicadas:", sucio.duplicated().sum())
sucio.drop_duplicates()
```

## 10. Ejemplo final: las ventas de una cafetería

Vamos a juntar todo lo aprendido con un pequeño análisis. Tenemos las ventas de tres productos durante tres meses:

```python
ventas = pd.DataFrame({
    "mes": ["Ene", "Ene", "Ene", "Feb", "Feb", "Feb", "Mar", "Mar", "Mar"],
    "producto": ["Café", "Té", "Bollo"] * 3,
    "unidades": [120, 40, 80, 135, 35, 95, 150, 30, 110],
    "precio": [1.2, 1.5, 1.8] * 3,
})

ventas
```

**1. Calcular los ingresos de cada venta** (unidades por precio):

```python
ventas["ingresos"] = (ventas["unidades"] * ventas["precio"]).round(2)
ventas
```

**2. Ingresos totales:**

```python
print("Ingresos totales:", round(ventas["ingresos"].sum(), 2), "euros")
```

**3. Ingresos de cada mes** (agrupando por mes):

```python
por_mes = ventas.groupby("mes")["ingresos"].sum()
por_mes
```

**4. ¿Qué producto genera más dinero?**

```python
por_producto = ventas.groupby("producto")["ingresos"].sum().sort_values(ascending=False)
print(por_producto)
print()
print("Producto más rentable:", por_producto.idxmax())
```

**5. Las 3 ventas individuales con más ingresos:**

```python
ventas.sort_values("ingresos", ascending=False).head(3)
```

**6. Guardar el resultado en un CSV:**

```python
ventas.to_csv("ventas_con_ingresos.csv", index=False)
print("Resultado guardado en ventas_con_ingresos.csv")
```
