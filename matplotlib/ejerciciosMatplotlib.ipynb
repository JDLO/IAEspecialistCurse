## 1. Ejercicio 2
Dibuja la evolución de la temperatura durante la semana con una línea roja, con un círculo en cada
punto y con cuadrícula. Pon el título "Temperaturas de la semana" y las etiquetas "Día" y "Grados".

Datos de partida:
> ```Python
dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]

`````python
dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]
temp = [18, 21, 19, 24, 22, 26, 25]

plt.plot(dias, temp, color="red", marker="o", linestyle="--")
plt.title("Temperaturas de la semana")
plt.xlabel("Días de la semana")
plt.ylabel("Temperatura (°C)")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
`````

## 3. Ejercicio 4
Dibuja el seno y el coseno de x en el mismo gráfico, en una figura de 8 x 4 pulgadas. El seno, con línea
continua azul; el coseno, con línea discontinua naranja. Añade leyenda y cuadrícula.
Datos de partida:
```Python
x = np.linspace(0, 2 * np.pi, 100)
```
*Pista: plt.figure(figsize=(8, 4)) va ANTES de dibujar. Para la línea discontinua: linestyle="--".*

`````python
x = np.linspace(0, 2 * np.pi, 100)
ySen = [np.sin(x) for x in x]
yCos = [np.cos(x) for x in x]

plt.figure(figsize=(8,4))
plt.plot(x, ySen, label="Seno", color='blue')
plt.plot(x, yCos, label="Coseno", color='red')
plt.legend(loc='lower left')
plt.grid(True)
plt.show()
`````

## 4. Ejercicio Edades

Realizar un diagrama de puntos donde se muestre cosas cheveres

`````python
filename = 'alturas_paises.csv'
data = np.genfromtxt(filename, delimiter=',', skip_header=1, dtype=None)
countries = np.genfromtxt(filename, delimiter=',', skip_header=1, usecols=(0), dtype=str)
countries = np.unique(countries)

listColor =['b','g','r','c','m','y','k','w']

colorDictionary = {countrie:listColor[i] for i, countrie in enumerate(countries)}

for countrie in countries:
    condition = data["f0"] == countrie
    plt.scatter(data[condition]['f0'],data[condition]['f1'],label=countrie, color=colorDictionary.get(countrie, 'gray'))

plt.title("Tabal de alturas")
plt.show()



`````

`````python
for countrie in countries:
    indexs = np.where(data['f0'] == countrie)
    plt.scatter(indexs,data[indexs]['f1'],label=countrie, color=colorDictionary.get(countrie, 'gray'), alpha=0.5)

plt.title("Tabal de alturas")
plt.show()
`````
