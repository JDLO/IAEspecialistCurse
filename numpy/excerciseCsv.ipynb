# Ejercicio de materias en **Numpy**

```python
import numpy as np

data = np.genfromtxt('notas.csv', delimiter=',', filling_values=0, skip_header=1,usecols=(1,2,3,4), dtype=int)
names = np.genfromtxt('notas.csv', delimiter=',', skip_header=1,usecols=(0), dtype=str)
subjects = np.genfromtxt('notas.csv', delimiter=',',max_rows=1, usecols=(1,2,3,4), dtype=str)
```

## Calculo de la media de todas las notas

```python
averageGrades = np.mean(data)
print(f"La media de las notas de todos los estudiantes son: {averageGrades}")
```

## Calcular la media de cada alumno

```python
averageGradesForStudent = np.mean(data, axis=1)
'''
for i, mean in enumerate(averageGradesForStudent):
    print(f"La media de notas de {names[i]} es {mean}")
'''

averageGradesStudens = np.array([names, averageGradesForStudent]).T
print(averageGradesStudens)
```

## Calcular la media de cada asignatura

```python
averageGradesSubjets = np.mean(data, axis=0)
averageGradeForSubjects = np.array([subjects, averageGradesSubjets]).T
print(averageGradeForSubjects)
```

## Encontrar la nota maxima y minima

```python
gradeMin = np.min(data)
gradeMax = np.max(data)

print(f"La nota minima es: {gradeMin}")
print(f"La nota maxima es:{gradeMax}")
```

## Encontrar el alumno con la mejor media de notas

```python
indice = np.argmax(averageGradesForStudent)
print(f"El estudiante con mejor media es: {names[indice]} con media de {np.max(averageGradesForStudent)}")
```
