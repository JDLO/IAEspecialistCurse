# Ejercicio de materias en **Numpy**

```python
import numpy as np

data = np.genfromtxt('notas.csv', delimiter=',', filling_values=0, skip_header=1,usecols=(1,2,3,4), dtype=int)
names = np.genfromtxt('notas.csv', delimiter=',', skip_header=1,usecols=(0), dtype=str)
subjects = np.genfromtxt('notas.csv', delimiter=',',max_rows=1, usecols=(1,2,3,4), dtype=str)
```

1. Calculo de la media de todas las notas

```python
averageGrades = np.mean(data)
print(f"La media de las notas de todos los estudiantes son: {averageGrades}")
```

2. Calcular la media de cada alumno

```python
averageGradesForStudent = np.mean(data, axis=1)
'''
for i, mean in enumerate(averageGradesForStudent):
    print(f"La media de notas de {names[i]} es {mean}")
'''

averageGradesStudens = np.array([names, averageGradesForStudent]).T
print(averageGradesStudens)
```

3. Calcular la media de cada asignatura

```python
averageGradesSubjets = np.mean(data, axis=0)
averageGradeForSubjects = np.array([subjects, averageGradesSubjets]).T
print(averageGradeForSubjects)
```

4. Encontrar la nota maxima y minima

```python
gradeMin = np.min(data)
gradeMax = np.max(data)

print(f"La nota minima es: {gradeMin}")
print(f"La nota maxima es:{gradeMax}")
```

5. Encontrar el alumno con la mejor media de notas

```python
indice = np.argmax(averageGradesForStudent)
print(f"El estudiante con mejor media es: {names[indice]} con media de {np.max(averageGradesForStudent)}")
```

6. Mostrar los alumnos aprobados, media mayor o igual a 5

```python
indexStudentsAproved = np.where(averageGradesForStudent >= 5)
#studentsAproved = np.array([names[indexStudentsAproved], averageGradesForStudent[indexStudentsAproved]]).T
studentsAproved = averageGradesStudens[indexStudentsAproved]

print(studentsAproved)
```

7. Econtrar todos los estudiantes que hayan aprovado todas las materias

```python
filas = np.all(data > 5, axis=1)
print(names[filas])
```

8. Econtrar cuantos estudiantes que aprobaron la media de la asignaturas y los que aprobaron todas las asignaturas

```python
countAverage = len(studentsAproved)
countSubjets = len(names[filas])
print(countAverage)
print(countSubjets)
```

9. Encontrar los alumnos que tienen 8 en Ruso

```python
studentsGrade8Ruso = np.where(data[:,3] == 8)
print(names[studentsGrade8Ruso])
```

10. Ordenar los alumnos por su media

```python
indexSortStudentsAverage = np.argsort(averageGradesForStudent)
indexReverseSortStudentAverage = np.argsort(averageGradesForStudent)[::-1]

print(*[f"El estudiante {student[0]}, obtuvo media de {student[1]}" for student in averageGradesStudens[indexSortStudentsAverage]], sep='\n')
print('-'*5,'ORDEN CONTRARIO', '-'*5)
print(*[f"El estudiante {student[0]}, obtuvo media de {student[1]}" for student in averageGradesStudens[indexReverseSortStudentAverage]], sep='\n')
```

11. Calcular la desviasion estandar

```python
standardDeviation = np.std(data, axis=0)
print(standardDeviation)
stadardDesviationBySubject = {subject: standardDeviation[i] for i, subject in enumerate(subjects) }
display(stadardDesviationBySubject)
```

12. use unique para eliminar datos repetidos

```python
result = np.unique(data, axis=0)
print(data)
print('-'*10, 'Datos limpios', '-'*10)
print(result)
print(f"lent of data {len(data)}, lent of result {len(result)}")
```
