# 🍕 Ejercicio KMeans — Clientes de comida a domicilio

Una app de comida a domicilio quiere **agrupar a sus clientes en tipos parecidos** para hacerles ofertas distintas.

El archivo `clientes_comida.csv` tiene 240 clientes con 3 columnas:
- `edad` (años)
- `pedidos_mes` (pedidos que hace al mes)
- `importe_medio` (euros que gasta de media en cada pedido)

## 1. Cargar los datos
Carga `clientes_comida.csv` en una variable `df`. Muestra las 5 primeras filas y cuántas filas y columnas tiene.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("./csv/clientes_comida.csv")
print(df.shape)
df.head()
```

✅ **Respuesta:** 240 filas y 3 columnas.

## 2. Ver los datos
Dibuja un gráfico de puntos con `pedidos_mes` en el eje X e `importe_medio` en el eje Y.

❓ ¿Cuántos grupos ves a simple vista?

```python
plt.scatter(df["pedidos_mes"], df["importe_medio"])
plt.xlabel("Pedidos al mes")
plt.ylabel("Importe medio (€)")
plt.show()
```

✅ **Respuesta:** se ven **4 nubes** de puntos.

## 3. Escalar
Escala todas las columnas y guarda el resultado en `X`.

```python
from sklearn.preprocessing import StandardScaler

escalador = StandardScaler()
X = escalador.fit_transform(df)
```

## 4. El método del codo
Calcula la inercia de KMeans para k de 1 a 10 y dibújala.

❓ ¿Qué número de grupos eliges?

```python
from sklearn.cluster import KMeans

inercias = [KMeans(n_clusters=k, random_state=42).fit(X).inertia_ for k in range(1, 11)]

plt.plot(range(1, 11), inercias, marker="o")
plt.xlabel("Número de grupos (k)")
plt.ylabel("Inercia")
plt.show()
```

✅ **Respuesta:** la línea se dobla en **k = 4**, así que elegimos 4 grupos.

## 5. Crear los grupos
Entrena KMeans con el número de grupos que has elegido. Guarda el grupo de cada cliente en una columna nueva `grupo` y repite el gráfico del paso 2 con un color por grupo.

```python
kmeans = KMeans(n_clusters=4, random_state=42)
df["grupo"] = kmeans.fit_predict(X)

plt.scatter(df["pedidos_mes"], df["importe_medio"], c=df["grupo"], cmap="tab10")
plt.xlabel("Pedidos al mes")
plt.ylabel("Importe medio (€)")
plt.show()
```

## 6. ¿Cuántos clientes hay en cada grupo?

```python
df["grupo"].value_counts()
```

## 7. ¿Cómo es cada grupo?
Calcula la media de cada columna por grupo.

❓ Ponle un nombre a cada grupo.

```python
df.groupby("grupo").mean().round(0)
```

✅ **Respuesta** (los números de grupo pueden salir en otro orden; fíjate en las medias):

| Cómo es el grupo | Nombre |
|---|---|
| ~22 años, ~12 pedidos, ~12 € | 🎓 **Estudiantes**: piden a menudo y barato |
| ~42 años, ~4 pedidos, ~46 € | 👨‍👩‍👧 **Familias**: piden poco pero mucha comida |
| ~33 años, ~19 pedidos, ~26 € | 💼 **Profesionales ocupados**: los que más piden |
| ~58 años, ~2 pedidos, ~20 € | 🕰️ **Ocasionales**: casi no usan la app |

## 8. Un cliente nuevo
Un cliente nuevo tiene **21 años**, hace **13 pedidos al mes** y gasta **11 €** de media. ¿A qué grupo pertenece?

```python
nuevo = pd.DataFrame({"edad": [21], "pedidos_mes": [13], "importe_medio": [11]})

print("Grupo:", kmeans.predict(escalador.transform(nuevo))[0])
```

✅ **Respuesta:** va al grupo de los **estudiantes** (joven, muchos pedidos baratos).

## 9. Para pensar
Repite el paso 5 con **k = 2**. ¿Qué grupos se han juntado? ¿Te sirven para hacer ofertas?

```python
kmeans2 = KMeans(n_clusters=2, random_state=42)
grupos2 = kmeans2.fit_predict(X)

plt.scatter(df["pedidos_mes"], df["importe_medio"], c=grupos2, cmap="tab10")
plt.xlabel("Pedidos al mes")
plt.ylabel("Importe medio (€)")
plt.show()
```

✅ **Respuesta:** con 2 grupos se mezclan perfiles muy distintos (por ejemplo, estudiantes con profesionales). Son demasiado generales para hacer ofertas pensadas para cada tipo de cliente.
