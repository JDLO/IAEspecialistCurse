# 🎮 Ejercicio KMeans: tipos de jugadores

Una empresa de videojuegos quiere conocer a sus jugadores para **hacerles ofertas distintas**. Tienes el archivo `jugadores.csv` con 250 jugadores y estas columnas:

| Columna | Qué significa |
|---|---|
| `jugador` | código del jugador |
| `horas_semana` | horas que juega a la semana |
| `gasto_mes` | euros que gasta al mes en el juego |
| `nivel` | nivel del personaje (1 a 100) |

**Tu misión:** usar KMeans para descubrir **qué tipos de jugadores hay** y ponerles nombre.

📁 **Antes de empezar:** sube `jugadores.csv` a Colab (icono de la carpeta 📁 de la izquierda → *Subir*).

## 1. Cargar los datos
Importa `pandas` y `matplotlib.pyplot`, lee `jugadores.csv` y muestra las primeras filas.

```python
import pandas as pd
import matplotlib.pyplot as plt

filename = './csv/jugadores.csv'

df = pd.read_csv(filename)

display(df.head())
```

## 2. Conocer los datos
¿Cuántas filas y columnas hay? Saca un resumen estadístico.

```python
display(df.describe())
```

## 3. Dibujar los datos
Haz un gráfico de puntos con `horas_semana` en el eje X y `gasto_mes` en el eje Y. Pon nombre a los ejes.

❓ ¿Cuántos grupos te parece ver a simple vista?

```python

plt.scatter(df["horas_semana"], df["gasto_mes"])
plt.show()
```

## 4. Elegir las columnas
Guarda en `X_datos` solo las columnas que sirven para medir parecidos entre jugadores.

```python
X_datos = df[["horas_semana", "gasto_mes", "nivel"]]
```

## 5. Escalar
KMeans mide distancias. Escala `X_datos` con `StandardScaler` y guarda el resultado en `X`.

❓ ¿Por qué hace falta?

```python
from sklearn.preprocessing import StandardScaler

escalador = StandardScaler()
X = escalador.fit_transform(X_datos)

display(X)
```

## 6. El método del codo
Calcula la inercia de KMeans para k de 1 a 10 y dibuja la gráfica.

❓ ¿Dónde está el codo? Ese será tu número de grupos.

```python
from sklearn.cluster import KMeans
rangeN = range(1, 11)

inercias = [KMeans(n_clusters=k, random_state=42).fit(X).inertia_ for k in rangeN]

plt.plot(rangeN, inercias, marker="o")
plt.xlabel("Numero de grupos (k)")
plt.ylabel("Inercia")
plt.show()
```

## 7. Hacer los grupos
Entrena KMeans con el número de grupos que has elegido, guarda el grupo de cada jugador en una columna nueva `grupo` y repite el gráfico del paso 3 **coloreado por grupo**.

```python
kmeans = KMeans(n_clusters=4, random_state=42)
df["grupo"] = kmeans.fit_predict(X)

plt.scatter(df["horas_semana"], df["gasto_mes"], c=df["grupo"], cmap="tab10")
plt.xlabel("Horas semana")
plt.ylabel("Gasto mes($)")
plt.show()
```

## 8. ¿Cuántos jugadores hay en cada grupo?

```python
display(df.groupby("grupo")["jugador"].count())
```

## 9. Describir los grupos
Calcula la **media** de las tres columnas numéricas en cada grupo.

```python
df.groupby("grupo")[["horas_semana", "gasto_mes", "nivel"]].mean().round(3)
```

Un jugador nuevo juega **25 horas** a la semana, gasta **80 €** al mes y tiene **nivel 90**. ¿A qué grupo va? ¿Qué oferta le harías?
